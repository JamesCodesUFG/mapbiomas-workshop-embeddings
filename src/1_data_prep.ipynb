```python
def scale_image(image):
    img_rescalada = ee.Image().byte()
    
    for _, bnd in enumerate(BANDS):
        amplitude = MIN_MAX_BAND_VALUES[bnd]['max'] - MIN_MAX_BAND_VALUES[bnd]['min']
        
        img_tmp = image.select(bnd).subtract(MIN_MAX_BAND_VALUES[bnd]['min']).divide(amplitude)
        img_tmp = img_tmp.multiply(255).toByte().rename(bnd)

        img_rescalada = img_rescalada.addBands(img_tmp)

    img_rescalada = img_rescalada.select(BANDS)

    return img_rescalada
```

```python
MOSAIC_YEAR = 2024

old_values = [0, 3, 4, 5, 6, 49, 11, 12, 29, 50, 13, 15, 19, 39, 20, 40, 62, 41, 46, 47, 35, 48,  9, 21, 23, 24, 75, 30, 25, 33, 31]
new_values = list(range(len(old_values)))

supervisedImg = (ee.Image('projects/mapbiomas-public/assets/brazil/lulc/collection10/mapbiomas_brazil_collection10_coverage_v2')
                      .select(f'classification_{MOSAIC_YEAR}')
                      .remap(old_values, new_values).unmask(0)
                      .rename(RESPONSE))

supervisedChannel = supervisedImg.toByte().rename(RESPONSE)

image = (ee.ImageCollection('projects/nexgenmap/MapBiomas2/LANDSAT/BRAZIL/mosaics-2')
            .filter(ee.Filter.eq('biome', 'CAATINGA'))
            .filter(ee.Filter.eq('year',  MOSAIC_YEAR))
            .mosaic())
        

image = scale_image(image.select(BANDS))
    
image = image.addBands(supervisedChannel)
```

```python
featureStack = image.select(BANDS + [RESPONSE]).unmask(0).float(),

list = ee.List.repeat(1, KERNEL_SIZE)
lists = ee.List.repeat(list, KERNEL_SIZE)
kernel = ee.Kernel.fixed(KERNEL_SIZE, KERNEL_SIZE, lists)

arrays = featureStack.neighborhoodToArray(kernel)
```

```python
ASSET_BIOMAS = 'projects/mapbiomas-workspace/AUXILIAR/biomas_IBGE_250mil'
ASSET_GRID_AMERICAS = 'projects/mapbiomas-raisg/PRODUCTOS/AGUA/DATOS_AUXILIARES/TEMP/sub_grid_0_25_america_sul'

def selec_set_polygons_train_test(feat_col):
    feat_col = feat_col.randomColumn('split_regions')
    feat_col = feat_col.randomColumn('split_samples')
    
    # selecionando a metade dos grides
    feat_col = feat_col.filter(ee.Filter.lt('split_regions', 0.01))
    
    # exportando os poligons de treino e Teste
    polg_train = feat_col.filter(ee.Filter.lt("split_samples", 0.7));  # 70% para traino
    polg_test = feat_col.filter(ee.Filter.gte("split_samples", 0.7));  # 30% para teste
    
    return polg_train, polg_test

nameBiomas = ["Cerrado", "Mata Atlântica", "Pampa", "Pantanal", "Amazônia", "Caatinga"]
bioma_name = nameBiomas[0]
bioma = (ee.FeatureCollection(ASSET_BIOMAS)
            .filter(ee.Filter.eq('Bioma', bioma_name))
            .geometry()
        )

subgrids = (ee.FeatureCollection(ASSET_GRID_AMERICAS)
                .filterBounds(bioma))

trainingPolys, validationPolys = selec_set_polygons_train_test(subgrids)
```

```python
polyImage = ee.Image(0).byte().paint(trainingPolys, 1).paint(validationPolys, 2)
polyImage = polyImage.updateMask(polyImage)

mapid = polyImage.getMapId({'min': 1, 'max': 2, 'palette': ['red', 'blue']})

map = folium.Map(location=[-1.3621, -45.2738], zoom_start=5)

folium.TileLayer(
    tiles=mapid['tile_fetcher'].url_format,
    attr='Google Earth Engine',
    overlay=True,
    name='supervisedLayer',
).add_to(map)

map.add_child(folium.LayerControl())

folium.TileLayer(
    tiles=mapid['tile_fetcher'].url_format,
    attr='Google Earth Engine',
    overlay=True,
    name='training polygons',
).add_to(map)

map
```

```python
RANDOM_SEED = 42

NUM_PIXELS_PER_POLYGON = 10

def export_to_drive(polys, numPixels, folder, fileNamePrefix):
  polysList = polys.toList(polys.size())

  for itter, poly in enumerate(polysList):
    geomSample = arrays.sample(
      region = poly.geometry(),
      scale = 30,
      numPixels = numPixels,
      seed = RANDOM_SEED
    )

    task = ee.batch.Export.table.toDrive(
      collection = geomSample,
      description = f"{fileNamePrefix}_g{itter}",
      folder = folder,
      fileNamePrefix = f"{fileNamePrefix}_g{itter}",
      fileFormat = 'TFRecord',
      selectors = FEATURES
    )
      
    task.start()

export_to_drive(trainingPolys, 10, GDRIVE_TRAIN_SAMPLES_DIR, "train_patches")
export_to_drive(validationPolys, 10, GDRIVE_VALID_SAMPLES_DIR, "valid_patches")
```
